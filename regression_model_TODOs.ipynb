```python
import pandas as pd
from matplotlib import pyplot as plt
from sklearn import metrics
from sklearn.metrics import ConfusionMatrixDisplay
from sklearn.linear_model import LogisticRegression

import import_ipynb
import preprocessing_data
```

```python
X_train = pd.read_csv("X_train_scaled.csv")
X_test  = pd.read_csv("X_test_scaled.csv")
y_train = pd.read_csv("y_train_scaled.csv")["Loan Status"]
y_test  = pd.read_csv("y_test_scaled.csv")["Loan Status"]

print("X_train:", X_train.shape, "X_test:", X_test.shape)
print("y_train:", y_train.shape, "y_test:", y_test.shape)
```

```python
# TO DO: Initialize the logistic regression model
model = LogisticRegression(penalty = "l2", C = 1.0, random_state = 16, solver='liblinear', max_iter=1000)
    
# TO DO: fit the model
model.fit(X_train, y_train)

#Can use model.coef_ or model.intercept_ for students to explore/visualize the regression model
```

```python
#TO DO: use predict_proba to get probabilities and make predictions
y_prob = model.predict_proba(X_test)[:, 1]   # probabilities in [0,1]
y_pred = (y_prob >= 0.5).astype(int)         # choose your threshold
```
